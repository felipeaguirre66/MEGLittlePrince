## Peek at the listed files

The next cell reads each path from `LISTEN_DIR` (configured in `config.py`). It displays text and structured files inline, embeds an audio player for the WAV, and includes a short purpose summary plus source for the Python and shell scripts. Missing paths are reported rather than stopping the rest of the preview.

```python
from pathlib import Path
import json
import sys

from IPython.display import Audio, Code, Markdown, display

# Find this project's config.py from the notebook's current working directory.
project_root = next(
    (directory for directory in (Path.cwd(), *Path.cwd().parents)
     if (directory / "config.py").is_file()),
    None,
)
if project_root is None:
    raise FileNotFoundError("Could not locate config.py in the current directory or its parents")
if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
from config import LISTEN_DIR

files_to_peek = [
    ".datalad/config",
    "datacite.yml",
    "dataset_description.json",
    "participants.json",
    "participants.tsv",
    "sourcedata/stimuli/audio/ch1-3.wav",
    "sourcedata/stimuli/data_path.txt",
    "sourcedata/stimuli/offset_stretch/infos.csv",
    "sourcedata/stimuli/origin.txt",
    "sourcedata/stimuli/phonemes/ch1-3.TextGrid",
    "sourcedata/stimuli/syntax/README.md",
    "sourcedata/stimuli/syntax/ch1-3.syntax.txt",
    "sourcedata/stimuli/syntax_new_no_punct/run0_v2_0.25_0.5-tokenized.syntax.txt",
    "sourcedata/stimuli/syntax_new_no_punct/run1_v2_0.25_0.5-tokenized.syntax.txt",
    "sourcedata/stimuli/syntax_new_untested/frtok.py",
    "sourcedata/stimuli/syntax_new_untested/make.sh",
    "sourcedata/stimuli/syntax_new_untested/run1_v2_0.25_0.5-tokenized.syntax.txt",
    "sourcedata/stimuli/syntax_new_untested/tsv_to_txt.py",
    "sourcedata/stimuli/txt_laser/run1.txt",
    "sourcedata/stimuli/txt_raw/ch1-3.txt",
    "sourcedata/task-listen_run-01_extra_info.tsv",
    "task-listen_events.json",
]

script_explanations = {
    "sourcedata/stimuli/syntax_new_untested/frtok.py": (
        "Tokenizes French text using Stanza. For each input file, it writes a "
        "sibling `-tokenized.txt` file with one tokenized sentence per line."
    ),
    "sourcedata/stimuli/syntax_new_untested/tsv_to_txt.py": (
        "Converts input TSV files into plain text by extracting the first "
        "column after the header, cleaning quotation marks and hyphens, and "
        "placing sentence-ending tokens at line breaks."
    ),
    "sourcedata/stimuli/syntax_new_untested/make.sh": (
        "Runs the preprocessing pipeline: TSV-to-text conversion, French "
        "tokenization, then Mind the Gap parsing to produce `.syntax.txt` files."
    ),
}


language_by_suffix = {
    ".json": "json",
    ".yml": "yaml",
    ".yaml": "yaml",
    ".py": "python",
    ".sh": "bash",
    ".tsv": "text",
    ".csv": "text",
    ".md": "markdown",
    ".txt": "text",
    ".textgrid": "text",
}

MAX_DISPLAY_LINES = 30
for relative_path in files_to_peek:
    path = Path(LISTEN_DIR) / relative_path
    display(Markdown(f"## `{relative_path}`"))
    if not path.is_file():
        display(Markdown(f"**Not found:** `{path}`"))
        continue

    if path.suffix.lower() == ".wav":
        display(Audio(filename=str(path)))
        continue

    content = path.read_text(encoding="utf-8", errors="replace")
    if path.suffix.lower() == ".json":
        try:
            content = json.dumps(json.loads(content), indent=2, ensure_ascii=False)
        except json.JSONDecodeError:
            pass

    if relative_path in script_explanations:
        display(Markdown(f"**What it does:** {script_explanations[relative_path]}"))

    lines = content.splitlines()
    if len(lines) > MAX_DISPLAY_LINES:
        omitted_lines = len(lines) - MAX_DISPLAY_LINES
        content = "\n".join(lines[:MAX_DISPLAY_LINES])
        content += f"\n… ({omitted_lines} more lines omitted)"

    display(Code(content, language=language_by_suffix.get(path.suffix.lower(), "text")))
```
